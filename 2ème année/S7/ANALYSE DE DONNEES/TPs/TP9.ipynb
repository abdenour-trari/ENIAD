```python
import pandas as pd 
import numpy as np 
```

```python
N = np.array([
        [20,10,5] ,
        [15,25,10],
        [5,10,20]])

n = N.sum()
F = N / n 

r = F.sum(axis=1)
c = F.sum(axis=0)

Dr = np.diag(1/np.sqrt(r))
Dc = np.diag(1/np.sqrt(c))

S = Dr @ (F - np.outer(r,c)) @ Dc

U, s, Vt = np.linalg.svd(S, full_matrices=False)

inertie = s**2
inerties_total = inertie.sum()

F_row = Dr @ U @ np.diag(s)
F_col = Dr @ Vt.T @ np.diag(s)

print(r , c)
print("+++++++++++++")
print(Dr , Dc)
print("+++++++++++++")
print(S)
print("+++++++++++++")
print(U,s,Vt)
print("+++++++++++++")
print(inertie,inerties_total)
print("+++++++++++++")
print(F_row,F_col)
```

```python
N = pd.DataFrame(
    [[40, 25, 5],
     [20, 35, 25],
     [5, 15, 30]],
    index=["Secondaire", "Licence", "Master+"],
    columns=["Ouvrier", "Employé", "Cadre"]
)

N
```

```python
r = P.sum(axis=1)
c = P.sum(axis=0)

r, c
```

```python
N_theo = n * np.outer(r, c)
N_theo = pd.DataFrame(N_theo, index=N.index, columns=N.columns)
N_theo
```

```python
chi2 = ((N - N_theo)**2 / N_theo).values.sum()
chi2
```

```python
ddl = (N.shape[0]-1)*(N.shape[1]-1)
ddl
```

```python
inertie_totale = chi2 / n
inertie_totale
```

```python
profils_lignes = P.div(r, axis=0)
profils_lignes
```

```python
rcT = np.outer(r, c)
M_cp = P - rcT
M_cp = pd.DataFrame(M_cp, index=N.index, columns=N.columns)
M_cp
```

```python
Dr_inv_sqrt = np.diag(1 / np.sqrt(r))
Dc_inv_sqrt = np.diag(1 / np.sqrt(c))

S = Dr_inv_sqrt @ M_cp.values @ Dc_inv_sqrt
S = pd.DataFrame(S, index=N.index, columns=N.columns)
S
```

```python
U, singular_values, Vt = np.linalg.svd(S, full_matrices=False)
```

```python
eigenvalues = singular_values**2
eigenvalues
```

```python
pourcentages = eigenvalues / eigenvalues.sum() * 100
pourcentages
```

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2
```

```python
N = pd.DataFrame(
    [[120, 30, 10],
     [60, 70, 30],
     [20, 40, 80]],
    index=["Capteur A", "Capteur B", "Capteur C"],
    columns=["Défaut faible", "Défaut moyen", "Défaut élevé"]
)

N
```

```python
n = N.values.sum()
n
```

```python
row_totals = N.sum(axis=1)
col_totals = N.sum(axis=0)

N_theo = np.outer(row_totals, col_totals) / n
N_theo = pd.DataFrame(N_theo, index=N.index, columns=N.columns)
N_theo
```

```python
chi2_stat = ((N - N_theo)**2 / N_theo).values.sum()
chi2_stat
```

```python
ddl = (N.shape[0] - 1) * (N.shape[1] - 1)
ddl
```

```python
chi2_crit = chi2.ppf(0.95, ddl)
chi2_crit
```

```python
chi2_stat > chi2_crit
```

```python
inertie_totale = chi2_stat / n
inertie_totale
```

```python
N = pd.DataFrame(
    [[25, 10, 5, 10],
     [30, 40, 10, 20],
     [20, 30, 25, 15],
     [10, 15, 20, 10]],
    index=["18-25", "26-35", "36-45", "46-55"],
    columns=["Produit A", "Produit B", "Produit C", "Produit D"]
)

N
```

```python
n = N.values.sum()
n
```

```python
F = N / n
F
```

```python
r = F.sum(axis=1)
c = F.sum(axis=0)

r, c
```

```python
Dr_inv_sqrt = np.diag(1 / np.sqrt(r))
Dc_inv_sqrt = np.diag(1 / np.sqrt(c))

rcT = np.outer(r, c)

S = Dr_inv_sqrt @ (F.values - rcT) @ Dc_inv_sqrt
S = pd.DataFrame(S, index=N.index, columns=N.columns)
S
```

```python
U, singular_values, Vt = np.linalg.svd(S, full_matrices=False)

eigenvalues = singular_values**2
eigenvalues
```

```python
pourcentages = eigenvalues / eigenvalues.sum() * 100
pourcentages
```

```python
FL = Dr_inv_sqrt @ U @ np.diag(singular_values)
FL = pd.DataFrame(FL, index=N.index)
FL
```

```python
FC = Dc_inv_sqrt @ Vt.T @ np.diag(singular_values)
FC = pd.DataFrame(FC, index=N.columns)
FC
```

```python
ctr_lignes = pd.DataFrame({
    "Axe 1": r.values * FL.iloc[:,0]**2 / eigenvalues[0],
    "Axe 2": r.values * FL.iloc[:,1]**2 / eigenvalues[1]
}, index=N.index)

ctr_lignes
```
