#### Exercice 1 :

```python
import numpy as np
```

```python
A = np.array([[1,2] , [3,4]])
B = np.array([[2,0] , [1,3]])
```

```python
print("A = \n" , A)
print("B = \n" ,B)
```

#### Exercice 2 :

```python
# Addition 
print(A + B)

# Soustraction 
print(A - B)

# (Produit élément par élément)
print(A * B)

# (produit matriciel)
print(A @ B)
```

#### Exercice 3 : 

```python
# Transposé d'une matrice : 
A_trans = np.transpose(A)
B_trans = np.transpose(B)


print("A = \n" , A)
print("B = \n" ,B)

print("Transposée de A : \n" , A_trans)
print("Transposée de B : \n" , B_trans)
```

```python
# Calclule de l'inverce d'une Matrice : 

det_A = np.linalg.det(A)
det_B = np.linalg.det(B)

if det_A != 0 :
    print("Inverse de A : \n" ,np.linalg.inv(A))
else :
    print("La matrice A n'est pas invercible !")

if det_B != 0 :
    print("Inverse de B : \n" ,np.linalg.inv(B))
else :
    print("La matrice B n'est pas invercible !")

```

```python
# Vérifiant B @ B-1 = I2


print(B @ np.linalg.inv(B)) 
```

#### Exercice 5 : 

```python
# Valeurs propres de A : 
 
x_A , vect_p_a = np.linalg.eig(A)

#pour les valeurs propres seulement on utlise : np.linalg.eig(M)

print(x_A[0],x_A[1])
```

```python
# vérifiant A * v = λ * v

print(np.allclose(A @ vect_p_a[: , 0] , x_A[0] * vect_p_a[: , 0])  ) 
```

#### Exercice 6 :

```python
C = np.array([[1 ,2 ,1],[0 ,1 ,0],[2 , 3 ,4]])
```

```python
print("C : \n" , C)
print("Transposée de C est : \n" , np.transpose(C))
print("Détermiant de C est : " , np.linalg.det(C))
if np.linalg.det(C) != 0 :
    print("L'inverse de C est : \n" , np.linalg.inv(C))
```

```python
# Valeurs propres de C : 
x_C = np.linalg.eigvals(C)
print("Les valeur propres de C sont : " , x_C[0] , x_C[1]) 
```

```python
# le rang d'une matrice : 
print("le rang de C est : " , np.linalg.matrix_rank(C))
```

```python
print(np.allclose(C @ np.linalg.inv(C) , np.eye(3)))
print(np.eye(3))
```
