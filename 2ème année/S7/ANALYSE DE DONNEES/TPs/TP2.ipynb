### Exercice 1 : 

```python
import numpy as np

a = np.array([[1],[1]])
a_t = np.transpose(a)

# calcul de a * a_t
numera = a @ a_t 

# calcul de a_t * a
dino = a_t @ a
```

```python
P = numera / dino
print(P)
```

```python
x = np.array([[2],[0]])
x0 = P @ x

print(x0)
```

```python
print(np.allclose(P@P , P))
print(np.allclose(np.transpose(P) , P))
```
