```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df = pd.read_excel("cars.xlsx")
df
```

```python
df_num = df.select_dtypes(include=['int64','float64'])

scaler = StandardScaler()
x_scaled = scaler.fit_transform(df_num) 
print(x_scaled)
```

```python
pca = PCA()
x_pca = pca.fit_transform(x_scaled)

print(x_pca)
```

```python

data = {
    "V": [85, 70, 90, 60, 80],
    "E": [90, 60, 85, 70, 75],
    "F": [80, 75, 95, 65, 85],
    "S": [70, 85, 60, 90, 75],
    "C": [75, 80, 65, 85, 70]
}

df = pd.DataFrame(data, index=["A1","A2","A3","A4","A5"])
df
```

```python
x_mean = df.mean()
df_c = df - x_mean

df_c
```

```python
X_c = np.cov(df_c)

print(X_c)
```

```python
val_p , vect_p = np.linalg.eig(X_c.T)
```
