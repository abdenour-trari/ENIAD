### Ex 1 

```python
import numpy as np

N = np.array([
    [20, 30, 50],
    [30, 20, 50],
    [50, 50, 100]
])

# 1. Calcul de la matrice des fréquences P
N_total = N.sum()
P = N / N_total
print("Matrice des fréquences P :\n", P)

# 2. Calcul des masses lignes r et colonnes c
r = P.sum(axis=1)  # somme sur les colonnes
c = P.sum(axis=0)  # somme sur les lignes
print("\nMasses lignes r :\n", r)
print("Masses colonnes c :\n", c)

# 3. Calcul de la matrice théorique d'indépendance N0
# N0 = total * r.T * c
N0 = np.outer(r, c)
print("\nMatrice théorique d'indépendance N0 :\n", N0)

# 4. Calcul de la matrice des résidus R
R = P - N0
print("\nMatrice des résidus R :\n", R)

# 5. Calcul de la distance du Khi-deux entre deux lignes
# Choisissons L1 et L2
# Formule : D^2(Li,Lj) = sum(( (P[Li] - P[Lj])**2 ) / c )
D2_L1_L2 = np.sum(( (P[0] - P[1])**2 ) / c.flatten())
print("\nDistance du Khi-deux entre L1 et L2 :\n", D2_L1_L2)
```

### Ex 2

```python
M = np.array([[10,5,5],
              [5,10,5],])

M_total = M.sum()
P = M / M_total
print("Matrice des fréquences P :\n", P)

c = np.sum(P , axis=0)
r = np.sum(P , axis=1)
print(c , r)

N0 = np.outer(r,c)
print('Matrice d\'independence : \n' , N0)

# Vérification de l'indépendance : calcul de la statistique du Khi-deux
khi2 = np.sum((P - N0)**2 / N0)
print("Statistique du Khi-deux : : " , khi2)

# Interprétation simple
if khi2 < 0.1 :
    print("Les variables semblement indépendantes")
else :
    print("Les variables ne semblent pas indépendantes.")
```

### Ex 5

```python
import numpy as np

# 1. Définir la matrice N (sans totaux)
N = np.array([
    [20, 15, 10, 5],
    [15, 20, 10, 5],
    [10, 10, 20, 10]
])

# Totaux lignes et colonnes
totaux_lignes = N.sum(axis=1)
totaux_colonnes = N.sum(axis=0)
total = N.sum()

# 1. Calcul des profils lignes fi = Nij / total ligne i
profiles_lignes = N / totaux_lignes[:, None]  # division ligne par ligne
print("Profils lignes (fi) :\n", profiles_lignes)

# 2. Calcul des masses colonnes cj = total colonne j / total général
c = totaux_colonnes / total
print("\nMasses colonnes (cj) :\n", c)

# 3. Calcul de toutes les distances du Khi-deux entre les lignes
n_lignes = N.shape[0]
distance_chi2 = np.zeros((n_lignes, n_lignes))

for i in range(n_lignes):
    for j in range(n_lignes):
        distance_chi2[i, j] = np.sum((profiles_lignes[i] - profiles_lignes[j])**2 / c)

print("\nMatrice des distances du Khi-deux entre les lignes :\n", distance_chi2)
```

```python
import numpy as np

# 1. Définir la matrice N
N = np.array([
    [20, 10, 20],
    [10, 30, 10],
    [20, 20, 10]
])

# Total général
total = N.sum()

# 1. Matrice des fréquences Pcv
P = N / total
print("Matrice des fréquences P :\n", P)

# 2. Masses lignes ri et colonnes cj
r = P.sum(axis=1)  # somme sur les colonnes
c = P.sum(axis=0)  # somme sur les lignes
print("\nMasses lignes ri :\n", r)
print("Masses colonnes cj :\n", c)

# 3. Profils lignes (fij = Nij / total ligne i)
profils_lignes = N / N.sum(axis=1)[:, None]
print("\nProfils lignes :\n", profils_lignes)

# 4. Distances du Khi-deux entre les lignes
# Formule : d^2(Lp,Lq) = sum_j ((fpj - fqj)^2 / cj)

# Ligne 1 et Ligne 2
d2_L1_L2 = np.sum((profils_lignes[0] - profils_lignes[1])**2 / c)
print("\nDistance du Khi-deux entre Ligne 1 et Ligne 2 :", d2_L1_L2)

# Ligne 1 et Ligne 3
d2_L1_L3 = np.sum((profils_lignes[0] - profils_lignes[2])**2 / c)
print("Distance du Khi-deux entre Ligne 1 et Ligne 3 :", d2_L1_L3)
```
