```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    'X1' : [4,2,6,8],
    'X2' : [2,0,3,6]
} , index= ['A' , 'B' , 'C' , 'D']
) 
df
```

```python
x_centred = df - df.mean()
x_centred
```

```python
# S = 1/n-1 * Xc^t * Xc

S = 1/3 * (np.transpose(x_centred) @ x_centred)
S = np.cov(x_centred.T)
print(S)
```

### Exercice 2 :

```python
S_s = np.cov(S.T)

val_p , vect_p = np.linalg.eig(S)

print(val_p)
print(vect_p)

totale_val = sum(val_p)
for i in val_p :
    print ("pour lambda :", i/totale_val)
```

### Exercice 3 :

```python
X = pd.DataFrame({
'X1': [2, 3, 4, 5],
'X2': [0, 1, 2, 3],
'X3': [1, 2, 3, 4]
})
X
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_scaler = scaler.fit_transform(X)

pca = PCA(n_components=3)
x_pca = pca.fit_transform(x_scaler)

x_var_ex = pca.explained_variance_ratio_
x_comp_prin = pca.components_


print(x_var_ex , x_comp_prin)
x_pca
```
