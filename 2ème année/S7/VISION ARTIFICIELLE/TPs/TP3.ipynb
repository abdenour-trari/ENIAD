```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imgA = cv2.imread("I1.jpg")
imgB = cv2.imread("I2.jpg")
assert imgA is not None and imgB is not None

h = min(imgA.shape[0], imgB.shape[0])
w = min(imgA.shape[1], imgB.shape[1])
A = cv2.resize(imgA, (w, h))
B = cv2.resize(imgB, (w, h))

blend = cv2.addWeighted(A, 0.7, B, 0.3, 0.9)

plt.imshow(cv2.cvtColor(blend, cv2.COLOR_BGR2RGB))
plt.title("Blending")
plt.axis("off")
plt.show()
```

```python
for k in range(0, 11):
    alpha = k / 10.0
    blend = cv2.addWeighted(A, 1 - alpha, B, alpha, 0.0)

    plt.imshow(cv2.cvtColor(blend, cv2.COLOR_BGR2RGB))
    plt.title(f"alpha = {alpha:.1f}")
    plt.axis("off")
    plt.show()
```

```python
h, w = imgA.shape[:2]
mask = np.zeros((h, w), dtype=np.uint8)

cv2.rectangle(mask, (w//4, h//4), (3*w//4, 3*h//4), 255, -1)

plt.imshow(mask, cmap="gray")
plt.title("Masque binaire")
plt.axis("off")
plt.show()
```

```python
roi = cv2.bitwise_and(imgA, imgA, mask=mask)

plt.imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB))
plt.title("Région d'intérêt")
plt.axis("off")
plt.show()
```

```python
mask_inv = cv2.bitwise_not(mask)
bg = cv2.bitwise_and(imgA, imgA, mask=mask_inv)

plt.imshow(mask_inv, cmap="gray")
plt.title("Masque inversé")
plt.axis("off")
plt.show()

plt.imshow(cv2.cvtColor(bg, cv2.COLOR_BGR2RGB))
plt.title("Fond")
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(imgA, cv2.COLOR_BGR2GRAY)
hist = cv2.calcHist([gray], [0], None, [256], [0, 256])

plt.imshow(gray, cmap="gray")
plt.title("Image grise")
plt.axis("off")
plt.show()

plt.plot(hist)
plt.title("Histogramme")
plt.xlabel("Intensité")
plt.ylabel("Nombre de pixels")
plt.show()
```
